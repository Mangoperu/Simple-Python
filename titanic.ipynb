```python
input_cols = list(data.columns)[1:-1]
target = 'Survived'
```

```python
print("input columns = " , input_cols)
print("target columns = " , target)
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(sparse_output = False , handle_unknown = 'ignore')
```

```python

data['Embarked'] = data['Embarked'].fillna("unknown")
data.info()
```

```python
encoder.fit(data[['Embarked']])
```

```python
encoder.categories_
```

```python
encoded_cols = encoder.get_feature_names_out(list(['Embarked']))
print(encoded_cols)
```

```python
data[encoded_cols] = encoder.transform(data[['Embarked']])
```

```python
data.head()
```

```python
data = data.drop(['Embarked']  , axis  =1)
```

```python
data.head()
all_cols = data.columns
```

```python

from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
data = scaler.fit_transform(data)
```

```python
data = pd.DataFrame(data  , columns = all_cols)
```

```python
data.head()
input_cols = data.columns[1:-1]
```

```python
data.describe()
```

```python
from sklearn.model_selection import train_test_split
```

```python
train_val, test_data = train_test_split(data , test_size=0.2  , random_state=42)
train_data , val_data = train_test_split(train_val , test_size=0.25  , random_state=42)
```

```python
train_inputs = train_data[input_cols].copy()
train_targets = train_data[target].copy()
```

```python
test_inputs = test_data[input_cols].copy()
test_targets = test_data[target].copy()
```

```python
val_inputs = val_data[input_cols].copy()
val_targets = val_data[target].copy()
```

```python
%%time

train_inputs = pd.read_parquet('train_inputs.parquet')
val_inputs = pd.read_parquet('val_inputs.parquet')
test_inputs = pd.read_parquet('test_inputs.parquet')

train_targets = pd.read_parquet('train_targets.parquet')[target]
val_targets = pd.read_parquet('val_targets.parquet')[target]
test_targets = pd.read_parquet('test_targets.parquet')[target]
```
